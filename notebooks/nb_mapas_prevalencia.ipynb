## Geogps Perú

```python
import geopandas as gpd
```

```python
shapefile_path = r"..\Departamental INEI 2023 geogpsperu SuyoPomalia\Departamental INEI 2023 geogpsperu SuyoPomalia.shp"
departamentos = gpd.read_file(shapefile_path)

print(departamentos.columns)
```

```python
cols_interes = [
    'DEPARTAMEN',
    'geometry'
]

departamentos = departamentos[cols_interes]
departamentos = departamentos.rename(columns={'DEPARTAMEN': 'departamento', 'geometry': 'geometry_region'})
```

```python
#convertir todos los valores de la columna departamento a minusculas
departamentos['departamento'] = departamentos['departamento'].str.lower()

#convertir los espacios en blanco a guiones bajos
departamentos['departamento'] = departamentos['departamento'].str.replace(' ', '_')

departamentos
```
